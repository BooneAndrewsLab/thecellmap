```python
d = Dataset.objects.get(pk=19)
```

```python
layout = json.load(open(d.static_path('layout.json')))['nodes']
```

```python
layout = {l['id']: (l['x'], l['y']) for l in layout}
```

```python
with open(d.static_path('nodes_inv.pickle')) as fp:
    nodes_inv = pickle.load(fp)
```

```python
nodes_inv_inv = {}
for nid, sids in nodes_inv.iteritems():
    for sid in sids:
        nodes_inv_inv[sid] = nid
```

```python
def reject_outliers(data, m = 2.):
    d = np.abs(data - np.median(data))
    mdev = np.median(d)
    s = d/mdev if mdev else 0.
    return data[s<m]
```

```python
rg = RegionGroup.objects.get(pk=48)
for term in Term.objects.filter(annotation=17):
    minx = maxx = miny = maxy = None
    xs = []
    ys = []
    for s in term.strains.select_related():
        if s.pk not in nodes_inv_inv: continue
        node = nodes_inv_inv[s.pk]
        x, y = layout[node]
        xs.append(x)
        ys.append(y)
        
        if minx is None:
            minx = maxx = x
            miny = maxy = y
        
        minx = min(minx,x)
        maxx = max(maxx,x)
        miny = min(miny,y)
        maxy = max(maxy,y)
    
    xs = list(reject_outliers(np.array(xs)))
    ys = list(reject_outliers(np.array(ys)))
    
    xr = abs(min(xs) - max(xs)) / 2
    yr = abs(min(ys) - max(ys)) / 2
    
    r, _ = Region.objects.get_or_create(name=term.name, alias=term.alias, region_group=rg, color=term.color)
    r.x = np.median(xs)
    r.y = np.median(ys)
    r.r = min(xr, yr)
    r.save()
```
