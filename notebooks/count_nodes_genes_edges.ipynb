```python
import json
```

```python
ds = Dataset.objects.get(pk=19) # Living dataset
```

```python
layout = json.load(open(ds.static_path('layout.json')))['nodes']
nodes = json.load(open(ds.static_path('nodes.json')))['nodes']
correlations = json.load(open(ds.static_path('correlations.json')))
```

```python
print("# Nodes: %d" % (len(layout), ))
```

```python
layout_ids = {x['id'] for x in layout}
nodes_in_layout = [n for n in nodes if n['id'] in layout_ids]
nodeids_in_layout = {n['id'] for n in nodes_in_layout}
```

```python
unique = {n['orf'] for n in nodes_in_layout}
print("# Unique genes: %d" % (len(unique), ))
```

```python
print("# Defined edges: %d" % (len(correlations['edges'])))
```

```python
used_edges = [e for e in correlations['edges'] if (e['s'] in nodeids_in_layout and e['t'] in nodeids_in_layout)]
print("# Shown edges: %d" % (len(used_edges)))
```
