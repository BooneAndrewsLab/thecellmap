```python
import pandas as p
import numpy as np
from collections import defaultdict

%pylab inline
```

```python
dataset = Dataset.objects.get(name='Living')
```

```python
xls = p.read_excel('/home/matej/Downloads/mc-1195618_table_s2.xls', sheetname='P<0.001')
```

```python
genes = {}
for g in Gene.objects.all():
    genes[g.orf] = g
    if g.name:
        genes[g.name] = g
    for a in g.aliases:
        if a not in genes:
            genes[a] = g
```

```python
arrays = dataset.arrays.select_related('gene')
queries = dataset.queries.select_related('gene')
data = p.DataFrame(np.full((queries.count(), arrays.count()), np.nan), index=queries, columns=arrays)
```

```python
querymap = {q.pk: q for q in queries}
```

```python
i = 1
for strain, scores in StrainData.objects.filter(dataset=dataset, type='Q').select_related('strain__gene').values_list('strain', 'scores'):
    data.loc[querymap[strain], :] = scores
    print(i, strain)
    i += 1
```

```python
arraygenes = defaultdict(list)
for a in arrays:
    arraygenes[a.gene].append(a)
```

```python
querygenes = defaultdict(list)
for a in queries:
    querygenes[a.gene].append(a)
```

```python
for _, x in xls.iterrows():
    g1 = x['GENE 1'].upper()
    g2 = x['GENE 2'].upper()
    
    g1name = g1
    g2name = g2
    
    if g1 not in genes or g2 not in genes:
        print(g1, g2, "\"One of the genes not found\"")
        continue
    
    g1 = genes[g1]
    g2 = genes[g2]
    
    ab = data.loc[querygenes[g1], arraygenes[g2]]
    ba = data.loc[querygenes[g2], arraygenes[g1]]
    
    if ab.empty and ba.empty:
        print(g1name, g2name, '"No pairs found"')
        continue
    
    ab = ab.unstack().dropna()
    ba = ba.unstack().dropna()
    
    abmax = np.nan
    bamax = np.nan
    
    if not ab.empty:
        abmax = ab[ab.abs().idxmax()]
    if not ba.empty:
        bamax = ba[ba.abs().idxmax()]
    
    print(g1name, g2name, abmax, bamax)
```
