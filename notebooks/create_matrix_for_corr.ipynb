```python
strain2node = {}
for nodeid, strains in pickle.load(open(dataset.static_path('nodes_inv.pickle'), 'rb')).items():
    for strain in strains:
        strain2node[strain] = nodeid
```

```python
data = []
index = []
for d in datas:
    data.append(d.scores)
    index.append(strain2node[d.strain_id])
```
