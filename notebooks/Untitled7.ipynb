```python
import json
```

```python
supp = json.load(open('/home/matej/S_collection_suppressors.json'))
nodes = json.load(open('../base/static/visualization/Living/nodes.json'))
genes = {}
for g in Gene.objects.all():
    genes[g.orf] = g
```

```python
node_supp = {}

for n in nodes['nodes']:
    if n['alel'] and n['alel'] in supp:
        suppressors = supp[n['alel']]
        
        clean = [tuple(s[2:]) for s in suppressors if s[2]]
        
        if not clean:
            continue
        
        node_supp[n['id']] = []
        for orf, state in set(clean):
            gene = genes[orf]
            node_supp[n['id']].append({'o': orf, 'n': gene.name or '', 's': state})
```

```python
len(node_supp)
```

```python
json.dump(node_supp, open('../base/static/visualization/Living/suppressors.json', 'w'), separators=(',', ':'))
```

```python
strain_supp = {}

for strain in Strain.objects.select_related('gene'):
    alel = strain.label().lower()
    if alel and alel in supp:
        suppressors = supp[alel]
        
        clean = [tuple(s[2:]) for s in suppressors if s[2]]
        
        if not clean:
            continue
        
        strain_supp[strain.pk] = []
        for orf, state in set(clean):
            gene = genes[orf]
            strain_supp[strain.pk].append({'o': orf, 'n': gene.name or '', 's': state})
```
