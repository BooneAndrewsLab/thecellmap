```python
from bs4 import BeautifulSoup
from urllib import urlopen
from PIL import Image
from StringIO import StringIO
import csv
```

```python
bs = BeautifulSoup(urlopen('http://live-donnelly-centre.pantheon.io/faculty'), 'html.parser')
```

```python
menu = bs.find_all("div", class_="secondary-menu")

with open('/home/yizhao/Downloads/pi_info.csv', 'w') as csvfile:
    writer = csv.writer(csvfile, delimiter=',',
                            quotechar='"', quoting=csv.QUOTE_MINIMAL)
    for pi in menu[0].find_all('a'):
        writer.writerow([pi.string, 'http://live-donnelly-centre.pantheon.io' + pi['href'], pi.string + '[Author] OR'])
```

```python
all_pi = Image.new('RGB', (len(menu[0].find_all('a'))*132, 198))
all_icon = Image.new('RGB', (len(menu[0].find_all('a'))*75, 75))

for index, pi in enumerate(menu[0].find_all('a')):
    bspi = BeautifulSoup(urlopen('http://live-donnelly-centre.pantheon.io' + pi['href']), 'html.parser')
    div = bspi.find_all('div', class_='profile_image')[0]
    src = div.find_all('img')[0]['src']
    
    img = Image.open(StringIO(urlopen(src).read())).resize((198, 198), Image.ANTIALIAS)
    
    pic = img.crop((33, 0, 165, 198))
    icon = img.resize((75, 75), Image.ANTIALIAS)
    
    all_pi.paste(pic, (index * 132, 0))
    all_icon.paste(icon, (index * 75, 0))

all_pi.save('/home/yizhao/all_pi.jpg')
all_icon.save('/home/yizhao/all_icon.jpg')
```
