```python
import pandas as p
from base.utils import gene_map
import numpy as np
```

```python
df = p.read_table('/home/matej/todo/mc_slim/go_slim_mapping.tab', header=None)
```

```python
g = df.groupby(3)
```

```python
genes = gene_map()
```

```python
for grp, gdata in g:
    annotation, isnew = Annotation.objects.get_or_create(
        name='GO Slim %s (2018-05-08)' % (grp, ),
        alias='go_slim_%s_20180508' % (grp.lower(), ),
        date='2018-05-08',
        version='G'
    )
    gdata[5].fillna('')
    print(annotation)
    
    terms = {}
    gene_map = {}
    
    for idx, row in gdata.iterrows():
        if row[0] not in genes: continue
        
        if row[4] not in terms:
            terms[row[4]], isnew = Term.objects.get_or_create(
                annotation=annotation,
                name=row[4],
                alias=row[5],
                source='',
            )
            print(terms[row[4]])
        
        gene_map.setdefault(row[4], []).append(genes[row[0]])
    
    for term, gns in gene_map.items():
        terms[term].genes.set(gns)
```
