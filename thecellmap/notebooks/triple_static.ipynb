```python
from django.db.models import Q
import json
import gzip
```

```python
genes = Gene.objects.filter(Q(strains1__isnull=False) | Q(strains2__isnull=False)).distinct()
```

```python
genes = list(genes)
```

```python
gene_obj = []
for gene in genes:
    go = {'id': gene.pk, 'o': gene.orf}
    if gene.name: go['n'] = gene.name
    if gene.aliases: go['a'] = gene.aliases
    
    gene_obj.append(go)
```

```python
with open('/home/matej/dev/workspace/thecellmap/trigenic/static/files/genes.json', 'w') as fout:
    json.dump(gene_obj, fout, separators=(',',':'))
```
