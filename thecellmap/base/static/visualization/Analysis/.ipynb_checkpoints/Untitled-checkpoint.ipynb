```python
import json
```

```python
layout = json.load(open('./layout.json'))
nodes = json.load(open('./nodes.json'))
```

```python
nodes = {n['id']: n for n in nodes['nodes']}
```

```python
coords = open('coords.csv', 'w')
for l in layout['nodes']:
    node = nodes[l['id']]
    coords.write('\t'.join(map(lambda x: x and str(x) or '', [node['orf'], node['name'], node['alel'], node['label'], l['x'], l['y']])) + '\n')

coords.close()
```
